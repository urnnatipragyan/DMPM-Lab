```python
3. # write a program to find the median of the data points without using function

n=int(input("enter how many data to be entered:"))
l=[int(input("enter a no.:"))for l in range(5)]
l.sort()
if (n%2==0):
  median = (l[n//2]+l[n//2-1])/2
else:
  median = l[n//2]
print("median is:",median)
```

```python
5. # write a program to find the mode of a datapoints without using function

n = int(input("Enter number of data points: "))
data = []
for i in range(n):
    x = int(input("Enter data: "))
    data.append(x)
mode = data[0]
max_count = 0
for i in data:
    count = 0
    for j in data:
        if i == j:
            count += 1
    if count > max_count:
        max_count = count
        mode = i
print("Mode =", mode)
```

```python
7. # wap to print varience of the data points without functions
n = int(input("Enter number of data points: "))
data = []
sum = 0
for i in range(n):
    x = float(input("Enter data: "))
    data.append(x)
    sum += x
mean = sum / n
variance = 0
for i in data:
    variance += (i - mean) ** 2
variance = variance / n
print("Variance =", variance)
```

```python
8. # wap to print the varience of data points with function
from statistics import variance

data_points = [10, 20, 30, 40, 50]

result = variance(data_points)

print("Variance is:", result)
```

```python
9. # wap to print the standard deviation with and without using function
import math
data = [10, 12, 23, 23, 16, 23, 21, 16]
total = sum(data)
mean = total / len(data)
sq_diff_sum = 0
for x in data:
    sq_diff_sum += (x - mean) ** 2
std_dev1 = math.sqrt(sq_diff_sum / (len(data) - 1))
print("Without Function:", std_dev1)
def get_std_dev(numbers):
    avg = sum(numbers) / len(numbers)
    sq_diff = sum((x - avg) ** 2 for x in numbers)
    return math.sqrt(sq_diff / (len(numbers) - 1))
std_dev2 = get_std_dev(data)
print("With Function:   ", std_dev2)
```
