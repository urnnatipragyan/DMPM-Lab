```python
import pandas as pd
import numpy as np
data=pd.read_csv("Titanic-Dataset.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
da=data.dropna()
```

```python
data.isnull().sum()
```

```python
avg=np.mean(data['Age'])
```

```python
data["Age"] = data["Age"].fillna(avg)
```

```python
# rename column name
data.rename(columns={"sex": "Gender"}, inplace=True)
```

```python
data['Embarked']=data['Embarked'].fillna(data['Embarked'].mode()[0])
```

```python
data.isnull().sum()
```

```python
data.drop(columns="Cabin",inplace=True)
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
import matplotlib.pyplot as plt
for i in data.select_dtypes([int, float]):
  plt.boxplot(data[i])
  plt.title(i)
  plt.show()
```

```python
for i in data.select_dtypes([int,float]):
  q1=data[i].quantile(0.25)
  q3=data[i].quantile(0.75)
  iqr=q3-q1
  lf=q1-(1.5*iqr)
  uf=q3+(1.5*iqr)
  data[i]=np.where(data[i]>uf,uf,data[i])
  data[i]=np.where(data[i]<lf,lf,data[i])
```

```python
for i in data.select_dtypes([int, float]):
  plt.boxplot(data[i])
  plt.title(i)
  plt.show()
```
