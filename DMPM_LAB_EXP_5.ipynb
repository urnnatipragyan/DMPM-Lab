```python
 # WRITE A PROGRAM TO IMPLEMENT Z-SCORE NORMALIZATION, MINMAX NORMALIZATION AND DECIMAL SCALING NORMALIZATION ON 10,20,30,40,50
import math
data = [10, 20, 30, 40, 50]
def min_max_normalization(X, new_min=0, new_max=1):
    min_val = min(X)
    max_val = max(X)
    return [ ((x - min_val) / (max_val - min_val))for x in X]
def z_score_normalization(X):
    mean = sum(X) / len(X)
    variance = sum((x - mean) ** 2 for x in X) / len(X)
    std_dev = math.sqrt(variance)
    return [(x - mean) / std_dev for x in X]
def decimal_scaling_normalization(X):
    max_abs = max(abs(x) for x in X)
    j = len(str(int(max_abs)))
    return [x / (10**j) for x in X]
print(f"Original Data: {data}\n")
print(f"Min-Max Normalized (0 to 1):{min_max_normalization(data)}")
print(f"Z-Score Normalized:{z_score_normalization(data)}")
print(f"Decimal Scaling Normalized:{decimal_scaling_normalization(data)}")
```

```python
# wap to implement the decimal scaling normalization on dataset from kaggle
import pandas as pd
df = pd.read_csv("Brand Awareness Publication.csv")
def decimal_scaling(series):
    max_val = abs(series.max())
    j = len(str(int(max_val)))
    return series / (10 ** j)
df["Cited by"] = df["Cited by"].fillna(0)
df["Cited_by_normalized"] = decimal_scaling(df["Cited by"])
print(df[["Cited by", "Cited_by_normalized"]].head(10))
```
