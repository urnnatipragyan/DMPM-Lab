```python
#calculate and analyse the corelation coefficent and covarient between 2 variable using any data.
import numpy as np
X = np.array([2, 3, 4, 5, 6])
Y = np.array([10, 15, 20, 25, 30])
x_mean = np.mean(X)
y_mean = np.mean(Y)
dx = X - x_mean
dy = Y - y_mean
cov = np.sum(dx * dy) / (len(X) - 1)
r = cov / (np.std(X, ddof=1) * np.std(Y, ddof=1))
print("Mean of X =", x_mean)
print("Mean of Y =", y_mean)
print("Covariance =", cov)
print("Correlation Coefficient =", r)
```
